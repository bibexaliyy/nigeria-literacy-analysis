```python
import pandas as pd

# If pandas isn't installed, run this in a separate cell: !pip install pandas

# adjust filename/path as needed (e.g., 'education_nga.csv' or full path)
df = pd.read_csv('education_nga.csv')
print(df.shape)
print(df.head())
print(df['Indicator Name'].nunique())
```

```python
literacy=df[df['Indicator Name'].str.contains('Literacy',case=False)]
print(literacy['Indicator Name'].unique())
```

```python
print(literacy.isnull().sum())
print(literacy['Year'].sort_values().unique())
print(literacy.dtypes)
 
```

```python
wide= literacy.pivot(index='Year',columns='Indicator Name',values='Value')
wide = wide.sort_index()
wide.to_csv('literacy_wide.csv')
```

```python
clean= wide.dropna()
clean.to_csv('literacy_wide_clean.csv')
print(clean)
```

```python
!pip install matplotlib
```

```python
import matplotlib.pyplot as plt

clean = clean.reset_index()
plt.plot(clean.index,clean['Literacy rate, adult female (% of females ages 15 and above)'],marker='o',label='Female')
plt.plot(clean.index,clean['Literacy rate, adult male (% of males ages 15 and above)'],marker='o',label='Male')
plt.xlabel('Year')
plt.ylabel('Literacy Rate (%)')
plt.title('Nigeria Adult Literacy Rate by Gender (1991-2024)')
clean['gender_gap'] = clean['Literacy rate, adult male (% of males ages 15 and above)'] - clean['Literacy rate, adult female (% of females ages 15 and above)']
print(clean[['Year', 'gender_gap']])
plt.legend()
plt.grid(True,alpha=0.3)
plt.show()
```
